## Quick brilliance transfer normalization
This notebook loads data from the Guide_BT_template instrument. If monitors named something with *sample* have a corresponding *source* version, that will be used to normalize bin for bin. Two new lists are created:
- normalized: List of monitor data objects that have been normalized
- references: List of the source data used to normalize

Its important to check the reference monitors have a flat distribution showing they are fully illuminated.

Change the "YOUR DATA HERE" string to a datafolder generated by the Guide_BT_template instrument.

McStasScript is used to load the data and plot the results.

```python
import mcstasscript as ms
```

```python
data = ms.load_data("Guide_BT_template_20251103_222039")
```

```python
import copy

normalized = []
references = []
for mon in data:
    if "sample" in mon.name:
        ref_name = mon.name.replace("sample", "source") 

        # find reference data
        ref_mon = ms.name_search(ref_name, data)
        references.append(ref_mon)

        # make copy to preserve the original data
        brilliance = copy.deepcopy(mon)

        # normalize the Intensity with the reference
        brilliance.Intensity = brilliance.Intensity/ref_mon.Intensity

        # change the label from Intensity to Brillinace if applicable
        if brilliance.metadata.ylabel == "Intensity":
            brilliance.set_ylabel("Brilliance")
        
        normalized.append(brilliance)
```

### Reference data, should be flat

```python
ms.make_sub_plot(references)
```

## Brilliance transfer data
Should have values between 0 and 1.

```python
ms.make_sub_plot(normalized)
```
